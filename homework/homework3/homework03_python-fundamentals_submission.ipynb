## 2. Compare Loop vs Vectorization

```python
# TODO: Compare speed using %timeit
def arr_add(arr, value):
    arr += value 
    return None
def arr_loop(arr, value):
    for i in range(len(arr)):
        arr[i] += value
    return None
%timeit arr + 1
%timeit for i in range(len(arr)): arr[i] += 1
%timeit arr_add(arr,1)
%timeit arr_loop(arr,1)
```
